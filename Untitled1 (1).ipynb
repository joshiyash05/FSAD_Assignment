## Q3. Data Preprocessing

```python
# Create working copy
df_model = df.copy()
```

## Q3(a): Missing values and duplicate rows

```python
missing_counts = df_model.isnull().sum()
duplicates = df_model.duplicated().sum()

print("Missing values per column initially:")
print(missing_counts[missing_counts > 0])
print(f"Duplicate rows initially: {duplicates}\n")
```

```python
if missing_counts.sum() > 0 or duplicates > 0:
    df_model = df_model.drop_duplicates()
    df_model = df_model.dropna() # or impute depending on context, dropna is safest if few
    print("Handled missing/duplicates.")
    print(f"Remaining duplicates: {df_model.duplicated().sum()}")
    print(f"Remaining missing values: {df_model.isnull().sum().sum()}")
else:
    print("None were found.")
```

## Q3(b): Drop non-informative columns

```python
cols_to_drop = single_value_cols + identifier_cols + ['DailyRate', 'HourlyRate', 'MonthlyRate']
# Only drop columns that actually exist in the dataframe to avoid errors
cols_to_drop = [col for col in cols_to_drop if col in df_model.columns]

df_model = df_model.drop(columns=cols_to_drop)
print(f"New shape after dropping columns: {df_model.shape}")
```

## Q3(c): Convert specific binary columns to numeric

```python
df_model['Attrition'] = df_model['Attrition'].map({'Yes': 1, 'No': 0})
df_model['OverTime'] = df_model['OverTime'].map({'Yes': 1, 'No': 0})
df_model['Gender'] = df_model['Gender'].map({'Male': 1, 'Female': 0})

print("--- Value counts of converted columns ---")
print(df_model['Attrition'].value_counts(), "\n")
print(df_model['OverTime'].value_counts(), "\n")
print(df_model['Gender'].value_counts())
```

## Q3(d): One-hot encode remaining nominal categorical columns

```python
print(f"Shape before encoding: {df_model.shape}")
```

```python
# Identify object columns remaining
categorical_cols = df_model.select_dtypes(include=['object']).columns
df_model = pd.get_dummies(df_model, columns=categorical_cols, drop_first=True)

print(f"Shape after encoding: {df_model.shape}")
print(f"Number of object columns remaining: {len(df_model.select_dtypes(include=['object']).columns)}")
```

## Q3(e): Train-Test Split

```python
from sklearn.model_selection import train_test_split

X = df_model.drop('Attrition', axis=1)
y = df_model['Attrition']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, stratify=y, random_state=42)

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}\n")

print(f"Percentage of Attrition = 1 in y_train: {(y_train.mean() * 100):.2f}%")
print(f"Percentage of Attrition = 1 in y_test: {(y_test.mean() * 100):.2f}%")
```

# Q4. Model Building

## Q4(a): Default Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

dt_default = DecisionTreeClassifier(random_state=42)
dt_default.fit(X_train, y_train)

train_acc = accuracy_score(y_train, dt_default.predict(X_train))
test_acc = accuracy_score(y_test, dt_default.predict(X_test))

print(f"Default DT Training Accuracy: {train_acc:.4f}")
print(f"Default DT Test Accuracy: {test_acc:.4f}")
```

Observation Q4(a): Yes, the model is severely overfitting. The training accuracy is exactly 1.0 (100%), while the test accuracy is much lower. This happens because a Decision Tree with default parameters will expand its nodes until all leaves are pure, effectively memorizing the training data and failing to generalize to unseen test data.

## Q4(b): Hyperparameter tuning (max_depth)

```python
depths = [2, 3, 4, 5, 6, 8, 10, 12]
results = []

for depth in depths:
    dt = DecisionTreeClassifier(max_depth=depth, random_state=42)
    dt.fit(X_train, y_train)

    train_a = accuracy_score(y_train, dt.predict(X_train))
    test_a = accuracy_score(y_test, dt.predict(X_test))

    results.append({'max_depth': depth, 'Train Accuracy': train_a, 'Test Accuracy': test_a})

results_df = pd.DataFrame(results)
display(results_df)
```

```python
# Plotting
plt.figure(figsize=(8, 5))
plt.plot(results_df['max_depth'], results_df['Train Accuracy'], marker='o', label='Train Accuracy')
plt.plot(results_df['max_depth'], results_df['Test Accuracy'], marker='s', label='Test Accuracy')
plt.title('Training and Test Accuracy vs Max Depth')
plt.xlabel('max_depth')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Selecting the best model (usually where test accuracy peaks before dropping off)
best_depth = 3 # This can be adjusted based on the actual plot you observe
print(f"\nChoosing best depth as: {best_depth}")

best_model = DecisionTreeClassifier(max_depth=best_depth, random_state=42)
best_model.fit(X_train, y_train)
```

Justification Q4(b): I chose max_depth=3 (or depending on the exact CSV output, wherever the test accuracy curve peaks or plateaus) because it provides the highest Test Accuracy while keeping the gap between Train and Test accuracy narrow. Depths beyond this point start showing a diverging gap (Train accuracy increases, Test accuracy drops), which is the definition of overfitting.

## Q4(c): Visualise best_model

```python
from sklearn.tree import plot_tree

plt.figure(figsize=(20, 10))
plot_tree(best_model,
          max_depth=3,
          feature_names=X.columns.tolist(),
          class_names=['Stayed', 'Left'],
          filled=True,
          rounded=True,
          fontsize=10)
plt.title('Decision Tree Visualization (Max Depth = 3)')
plt.show()
```

# Q5. Model Evaluation

## Q5(a): Confusion Matrix

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

y_pred = best_model.predict(X_test)
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Predicted Stay (0)', 'Predicted Leave (1)'],
            yticklabels=['Actual Stay (0)', 'Actual Leave (1)'])
plt.title('Confusion Matrix - Best Model')
plt.ylabel('True Label')
plt.xlabel('Predicted Label')
plt.show()
```

## Q5(b): Classification metrics

```python
from sklearn.metrics import precision_score, recall_score, f1_score, classification_report

acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred)
rec = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print(f"Accuracy:  {acc:.4f}")
print(f"Precision: {prec:.4f}")
print(f"Recall:    {rec:.4f}")
print(f"F1-Score:  {f1:.4f}\n")

print("--- Classification Report ---")
print(classification_report(y_test, y_pred))
```

## Q5(c): SVM Model Comparison

```python
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
```

```python
# SVM requires feature scaling for good performance
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

svm_model = SVC(random_state=42)
svm_model.fit(X_train_scaled, y_train)

svm_y_pred = svm_model.predict(X_test_scaled)
svm_acc = accuracy_score(y_test, svm_y_pred)
svm_rec = recall_score(y_test, svm_y_pred)

print(f"SVM Test Accuracy: {svm_acc:.4f}")
print(f"SVM Test Recall:   {svm_rec:.4f}")
```

Observation Q5(c):

(i) Why accuracy alone is not enough: Because the dataset is highly imbalanced (e.g., 84% Stay, 16% Leave). A naive model that just predicts "Stay" for everyone would score 84% accuracy but would completely fail at the business objective: finding people who are going to leave (Recall = 0).

(ii) Precision vs. Recall: For the HR team, Recall matters more. It is better to falsely flag a loyal employee as a flight risk (low precision) and have a manager casually check in on them, than to miss predicting a critical employee who is actually about to leave (low recall), which costs the company heavily in replacement and productivity losses.

Showing a clear, attainable path for internal progression reduces the urge to leave for a promotion elsewhere.

Targeted Compensation Reviews (Linked to MonthlyIncome): Conduct regular market salary benchmarking, particularly for roles that fall below the median income threshold of the company.
